# Matriz de confusion

![matriz confusion](https://img.grepmed.com/uploads/8345/specificity-table-confusionmatrix-biostatistics-contingency-original.jpeg)

# Contenido de la matriz 
 Supongamos que tenemos dos clases A, B --- **Número de columnas = Número de clases**
- **Verdadero Positivo**: Número de casos pertenece a **A** y que en realidad pertenece a **A**
- **Falso negativo**: Número de casos que NO pertenece a **A** y en realidad pertenece a **A**
- **Verdadero Negativo** Número de casos que pertenece a **B** y en realidad pertenece a **B**
- **Falso negativo**: Número de casos que NO pertenece a **B** y en realidad pertenece a **B**
- Y así sucesivamente, si tenemos más clases ....

# Sub-Términos
- **Precisión** = Valor predictivo positivo: La proporción de casos positivos que fueron predichos correctamente por el algoritmo = T(**A**) / T**A** + F**A**.
- **Recall** = Sensibilidad: La proporción de muestras predichas correctamente, para todos los casos que realmente pertenecen a esta clase en **todos los conjuntos de datos** = T (**A**) / T **A** + F(**B**)]. Porque **F(B) == T(A)**

Recordando que puede detectar la sensibilidad del algoritmo con cada clase.
- **Precisión**: la proporción del número total de predicciones que fueron correctas = **T (A) + T (B) / Población total**

- Especificidad: la proporción de casos negativos reales que se identifican correctamente.

# Beneficios 

La matriz de confusión no solo nos proporciona los errores cometidos por nuestro modelo de clasificación, sino también los tipos de errores que cometimos.

- **Alto recall y baja precisión**: representa que la mayoría de las predicciones positivas se reconocen correctamente
> Significa que tenemos menos falsos negativos sin embargo hay un aumento significativo en falsos positivos.
- **Recall baja y alta precisión** representa que nos perdemos muchos ejemplos positivos que en la vuelta tienen un alto (falso negativo), pero aquellos que predecimos como positivos son seguramente predicciones positivas.
- Una puntuación de **alta precisión** da más confianza a la capacidad del modelo para clasificar **(A)**. Combinar esto con **recall** da una idea de cuántos del total **(A)**s fue capaz de cubrir.

En el caso de las organizaciones de atención médica, estarán más preocupadas por un diagnóstico positivo erróneo mínimo.

Estarán más centrados en una alta especificidad. Por otro lado, otro modelo predictivo estará más preocupado por la sensibilidad.

# Area bajo la curva (AUC)

![title](https://images.squarespace-cdn.com/content/5e82b422e6ba6143dff54d04/1587138988168-ZI0B4RRCACX0XZEM6CYM/Posterior+probabilities+ROC+curve.JPG?content-type=image%2Fjpeg)

AUC es una métrica de rendimiento para problemas de clasificación binaria. El AUC representa la capacidad de un modelo para discriminar entre clases positivas y negativas.
- Un área de 1.0 representa un modelo que hizo todas las predicciones a la perfección. Un área de
- 0.5 representa un modelo que es tan bueno como aleatorio. La ROC se puede dividir en sensibilidad y especificidad.

Un problema de clasificación binaria es en realidad un compromiso entre sensibilidad y especificidad.
La sensibilidad es la verdadera tasa positiva, también llamada recuperación. Es el número de instancias de la clase positiva (primera) que realmente predijo correctamente. La especificidad también se denomina tasa negativa verdadera. Es el número de instancias de la clase negativa (segunda) que en realidad se predijeron correctamente.

- Las curvas ROC también son importantes, ya que es ideal para maximizar la tasa de verdaderos positivos y minimizar la tasa de falsos positivos.

# Beneficios
- Las curvas de los diferentes modelos de aprendizaje automático se pueden verificar con diferentes umbrales.
- La capacidad predictiva del modelo se resume en el área bajo la curva (AUC).
- El AUC se considera una variante escalada, mide el rango de las predicciones en lugar de sus valores absolutos.
- AUC siempre se centra en la calidad de las habilidades del modelo en la predicción, independientemente del umbral elegido.

## Precisión de clasificación
- La precisión de la clasificación es el número de predicciones correctas realizadas como una proporción de todas las predicciones realizadas.
- Esta es la métrica de evaluación más común para los problemas de clasificación, también es la más utilizada.
- Realmente solo es adecuado cuando hay un número igual de observaciones en cada clase (lo que rara vez ocurre) y que todas las predicciones y errores de predicción son igualmente importantes.

Se informa la relación. Esto se puede convertir en un porcentaje multiplicando el valor por 100, lo que da una puntuación de precisión de aproximadamente XX% de precisión.

## Reporte clasificacion

![title](https://www.researchgate.net/profile/Diego_Heras/publication/321176883/figure/fig12/AS:611805709746189@1522877256660/Reporte-de-clasificacion-de-Random-Forest.png)

# Contenido
- **Precisión** = Valor predictivo positivo: La proporción de casos positivos que fueron predichos correctamente por el algoritmo.
- **Recall** = Sensibilidad: La proporción de muestras predichas correctamente, para todos los casos que realmente pertenecen a esta clase en **todos los conjuntos de datos**.
- **F1-Score**: = **(2 * Precision * Recall) / (Precision + Recall)** 

# F1-Score

Media armónica de precisión y valores de recuperación. La fórmula para la puntuación de F1 es así:

**Cuanto mayor sea la puntuación F1, mayor será el poder predictivo del modelo de clasificación.**

- Una puntuación cercana a 1 significa un modelo perfecto,
- sin embargo, una puntuación cercana a 0 muestra una disminución en la capacidad predictiva del modelo

F1 Score es necesario cuando desea buscar un equilibrio entre Precision y Recall.

# Error medio cuadratico

- las métricas más populares utilizadas en problemas de regresión.
- RMSE se define por la desviación estándar de los errores de predicción.
- Estos errores de predicción a veces se denominan residuales.
- Los residuos son básicamente la medida de la distancia de los puntos de datos desde la línea de regresión.

![title](https://live.staticflickr.com/7903/32110444487_25fdaf9d52_b.jpg)

- Con RMSE se asume que los residuos son insesgados y siguen una **distribución normal**.
- Antes de usar RMSE, asegúrese de que no haya valores atípicos en el conjunto de datos porque RMSE está muy influenciado por valores atípicos.
- El error cuadrático medio tiene una ponderación más alta y también penaliza los errores en comparación con otras métricas de evaluación.


# Error absoluto medio

- El promedio tomado entre los valores originales y los valores predichos
- También mide la magnitud promedio del error, es decir, qué tan lejos están las predicciones de la salida real.
- Además, MAE no nos proporciona ninguna dirección de error, es decir, si estamos sobreajustando los datos o no ajustando los datos.

![title](https://ingenioempresa.com/wp-content/uploads/2016/02/formula-desviacion-media-absoluta.png)

# Error cuadratico medio

- MSE toma el promedio del cuadrado de la diferencia entre los valores originales y los valores predichos.
- En MSE, el cálculo de gradientes se vuelve más fácil que MAE, que requiere herramientas computacionales para calcular gradientes.
- El error cuadrático medio es útil cuando la columna de destino está **normalmente distribuida alrededor del valor medio**.
- El error cuadrático medio entra en escena cuando los valores atípicos están presentes en nuestro conjunto de datos y se hace necesario penalizarlos.

![title](https://www.i2tutorials.com/wp-content/media/2019/11/Mean-Bias-Error-1-i2tutorials-2.jpg)

# R cuadrado

- es una medida estadística de qué tan cerca se ajusta el punto de datos a la línea de regresión.
- También se conoce como el coeficiente de determinación. R-Cuadrado se define por la variación explicada dividida por la variación total que es explicada por el modelo lineal.


Este es un valor entre 0 y 1 para no ajuste y ajuste perfecto respectivamente.

# Ejemplo - Predicción de supervivencia

Usando el dataset de los pasajeros del Titanic, predecir quién hubiera sobrevivido.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.info()
```

```python
from sklearn.preprocessing import LabelEncoder

LE = LabelEncoder()
titanic['genero']= LE.fit_transform(titanic['sex'])
titanic['embarcado']= LE.fit_transform([s if type(s)==str else 'Z' for s in titanic['embarked']])
titanic['clase']= LE.fit_transform([s if type(s)==str else 'Z' for s in titanic['class']])

titanic.head()
```

```python
titanic = titanic[['survived','pclass','age','sibsp','parch','fare','clase','genero','embarcado','clase']]
titanic.head()
```

```python
titanic.describe()
```

```python
titanic.info()
```

```python
titanic = titanic.fillna(0)
```

```python
# Run a simple linear classifier    

from sklearn.metrics import roc_curve
from sklearn.metrics import auc

#   - Split-out validation dataset
#--------------------------------------------------------------------------------
array = titanic.values
X = array[:,1:] # All features without target
Y = array[:,0] # The Output we need to predict
validation_size = 0.20 # ex.
seed = 7
X_train, X_validation, Y_train, Y_validation = train_test_split(X, Y, test_size=validation_size, random_state=seed)

#---------------------------------------------------------------------------------

# - Run your Classification algorithm
clf = LogisticRegression(solver='lbfgs') 
clf.fit(X_train, Y_train)
y_preds = clf.predict_proba(X_validation)
# --------------------------------------------------------------------------------

# Calculate AUC
fpr, tpr, thresholds = roc_curve(Y_validation, y_preds[:,1])    
roc_auc = auc(fpr, tpr)
#---------------------------------------------------------------------------------

# Plot AUC
fig = plt.figure(figsize =(16,8))
ax2 = plt.axes()
ax2.plot(fpr, tpr, lw=1, label='(AUC = %0.2f)' % (roc_auc))
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('************Title***********')
plt.legend(loc="lower right")
plt.show()
```
